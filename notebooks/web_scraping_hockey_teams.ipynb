```python
from bs4 import BeautifulSoup
import requests
```

```python
url=requests.get('https://www.scrapethissite.com/pages/forms/?page_num=6')
```

```python
The_page=BeautifulSoup(url.text,'html')
```

```python
print(The_page)
```

```python
table_headers=The_page.find_all('th')
```

```python
print(table_headers)
```

```python
headers=[header.text.strip()for header in table_headers]
```

```python
print(headers)
```

```python
import pandas as pd
df=pd.DataFrame(columns=headers)
```

```python
df
```

```python
table = The_page.find('table')
```

```python
rows = table.find_all('tr')[1:]

for row in rows:
    table_final_data = [
        data.get_text(strip=True)
        for data in row.find_all('td')
    ]

    df.loc[len(df)] = table_final_data
```
